## 🔹 Training Input / Setup Terms

* **Epoch**
  One complete pass of the training dataset through the model.

* **Batch Size**
  Number of images processed together before updating model weights.

* **Image Size (imgsz)**
  Resolution of input images (e.g., 640×640).

* **Learning Rate (lr)**
  Controls how much model weights change during training.

* **Optimizer**
  Algorithm used to update weights (e.g., SGD, Adam).

* **Train / Val Dataset**

  * **Train**: Used to learn
  * **Val**: Used to evaluate performance after each epoch

---

## 🔹 Output / Loss Terms (Very Important)

These appear in training logs like:

```
box_loss  cls_loss  dfl_loss
```

### 1️ **Box Loss (box_loss)**

➡ Measures **how accurately bounding boxes are predicted**

* Compares predicted box vs actual box
* Based on **IoU (Intersection over Union)**
* **Lower = better localization**

 Example:
If the model draws boxes slightly off the object → box_loss increases

---

### 2️ **Classification Loss (cls_loss)**

➡ Measures **how correctly object classes are predicted**

* Used when multiple object classes exist
* Penalizes wrong class predictions
* **Lower = better classification**

 Example:
Predicting *cat* instead of *dog* → cls_loss increases

---

### 3️ **DFL Loss / DF1 Loss (dfl_loss / df1_loss)**

➡ **Distribution Focal Loss** (YOLOv8+)

* Improves **bounding box precision**
* Predicts box edges as probability distributions instead of fixed values
* Helps with **small and overlapping objects**

 Think of it as:

> “Fine-tuning box edges more precisely”

 Sometimes shown as:

* `dfl_loss`
* `df1_loss` (same concept, different naming)

---

## 🔹 Evaluation / Metric Terms

### 4️ **Precision**

➡ Out of all predicted objects, how many were correct?

```
Precision = Correct Predictions / Total Predictions
```

---

### 5️ **Recall**

➡ Out of all real objects, how many were detected?

```
Recall = Correct Predictions / Actual Objects
```

---

### 6️ **mAP (mean Average Precision)**

➡ Main accuracy metric for object detection

* **mAP@0.5** → IoU ≥ 50%
* **mAP@0.5:0.95** → Stricter, more realistic accuracy

 Higher mAP = better model

---

## 🔹 Training Behavior Terms

* **Overfitting**
  Model performs well on training data but poorly on validation data.

* **Underfitting**
  Model performs poorly on both training and validation.

* **Convergence**
  Loss values stabilize → model has learned well.

---

| cls_loss       | Class prediction accuracy   |
| dfl / df1_loss | Advanced box edge precision |

| mAP            | Overall detection accuracy  |
